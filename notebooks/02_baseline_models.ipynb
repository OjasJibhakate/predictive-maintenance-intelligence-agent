# 02 - Baseline models
Leakage-free comparison: Logistic Regression vs Random Forest vs XGBoost (3-fold CV, recall/precision/F1).

```python
import os, sys
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd
from sklearn.model_selection import train_test_split
from src.config import RAW_FEATURES, TARGET, RANDOM_SEED
from src.train import compare_models

df = pd.read_csv('../data/ai4i2020.csv').dropna(subset=RAW_FEATURES + [TARGET])
X = df[RAW_FEATURES]; y = df[TARGET].astype(int)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, stratify=y, random_state=RANDOM_SEED)
results = compare_models(X_tr, y_tr)
for name, s in results.items():
    print(f"{name}: recall={s['recall']:.3f} precision={s['precision']:.3f} f1={s['f1']:.3f}")
```
